```python
# sns.boxplot(x=df['MonthlyCharges'], y=df['Churn'])
# plt.title("Monthly Charges vs Churn")
# plt.show()/

plt.figure(figsize=(7,5))

sns.boxplot(
    data=df,
    x='Churn',
    y='MonthlyCharges',
    palette='Set3'
)

plt.title("Monthly Charges vs Customer Churn", fontsize=16, fontweight='bold')
plt.show()
```

```python
# sns.histplot(df['MonthlyCharges'],kde=True,palette='Set2')
# plt.title("Monthly Charges Distribution")
# plt.show()

plt.figure(figsize=(8,5))

sns.histplot(
    df['MonthlyCharges'],
    bins=30,
    kde=True,
    color='steelblue'
)

plt.title("Distribution of Monthly Charges", fontsize=16, fontweight='bold')
plt.xlabel("Monthly Charges")
plt.grid(alpha=.3)
plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x='Churn',
    hue='Churn',
    palette='Set2',
    legend=False
)


plt.title("Overall Customer Churn Distribution", fontsize=16, fontweight='bold')
plt.xlabel("Customer Status")
plt.ylabel("Number of Customers")
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(8,5))

ax = sns.countplot(
    data=df,
    x='gender',
    hue='Churn',
    palette='Set2'
)

for container in ax.containers:
    ax.bar_label(container)

plt.title(
    "Customer Churn Distribution by Gender",
    fontsize=16,
    fontweight='bold'
)

plt.xlabel("Gender", fontsize=12)
plt.ylabel("Number of Customers", fontsize=12)

plt.legend(title="Churn Status")

plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.tight_layout()

plt.show()
```

One gender shows a slightly higher churn rate than the other. Although the difference is not substantial, it may be worth investigating whether customer preferences or service usage patterns contribute to this variation.

```python
plt.figure(figsize=(8,5))

ax = sns.countplot(
    data=df,
    x='Contract',
    hue='Churn',
    palette='viridis'
)

for container in ax.containers:
    ax.bar_label(container)

plt.title("Customer Churn by Contract Type", fontsize=16, fontweight='bold')
plt.xlabel("Contract Type")
plt.ylabel("Customers")
plt.legend(title="Churn")
plt.grid(axis='y', alpha=.3)
plt.show()
```

```python
sns.set_style("whitegrid")
g = sns.pairplot(
    df[['tenure',
        'MonthlyCharges',
        'TotalCharges',
        'Churn']],
    hue='Churn',
    palette='Set2',
    diag_kind='kde',
    height=3,
    corner=True,
    plot_kws={
        'alpha':0.7,
        's':40
    }
)
g.fig.suptitle(
    "Relationship Between Numerical Features and Customer Churn",
    fontsize=18,
    fontweight='bold',
    y=1.02
)
plt.tight_layout()
plt.show()
```

```python
# sns.violinplot(data=df,
#                x='Churn',
#                y='MonthlyCharges')

plt.figure(figsize=(8,6))

sns.violinplot(
    data=df,
    x='Churn',
    y='MonthlyCharges',
    hue='Churn',
    palette='Set2',
    inner='quart',
    linewidth=1.5,
    legend=False
)

plt.title("Monthly Charges Distribution by Customer Churn",
          fontsize=16,
          fontweight='bold')
plt.xlabel("Customer Churn",
           fontsize=12)
plt.ylabel("Monthly Charges ($)",
           fontsize=12)
plt.grid(axis='y',
         linestyle='--',
         alpha=0.4)
plt.tight_layout()
plt.show()
```

Customers who churn generally have higher monthly charges than customers who stay. The wider shape of the violin indicates where most customers are concentrated, while the quartile lines show the spread of the data. This suggests that higher monthly pricing may contribute to customer churn.

```python
df['Charge_Category'] = pd.cut(
    df['MonthlyCharges'],
    bins=[0,35,70,120],
    labels=['Low','Medium','High']
)
```

```python
pd.crosstab(
    df['Charge_Category'],
    df['Churn'],
    normalize='index'
) * 100
```

```python
plt.figure(figsize=(8,5))

ax = sns.countplot(
    data=df,
    x='Charge_Category',
    hue='Churn',
    palette='crest'
)

for c in ax.containers:
    ax.bar_label(c)

plt.title("Monthly Charge Category vs Churn", fontsize=16, fontweight='bold')
plt.show()
```

"Customers in the high monthly charge category exhibit a greater tendency to churn, suggesting that pricing plays a role in customer decisions."

```python
df['Tenure_Group'] = pd.cut(
    df['tenure'],
    bins=[0,12,24,48,72],
    labels=['0-1 Year',
            '1-2 Years',
            '2-4 Years',
            '4-6 Years']
)
```

```python
tenure_churn = (
    pd.crosstab(
        df['Tenure_Group'],
        df['Churn'],
        normalize='index'
    ) * 100
)

ax = tenure_churn.plot(
    kind='bar',
    figsize=(9,5),
    colormap='Set2'
)

# Add percentage labels
for container in ax.containers:
    ax.bar_label(container, fmt='%.1f%%')

plt.title(
    "Customer Churn Rate by Tenure Group",
    fontsize=16,
    fontweight='bold'
)

plt.xlabel("Customer Tenure")
plt.ylabel("Percentage (%)")

plt.legend(title="Churn Status")

plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.tight_layout()

plt.show()
```

New customers are at the greatest risk of leaving the company. Customers who remain beyond the first year are significantly more likely to stay long-term, indicating that the early stages of the customer journey are critical for retention.

```python
services = ['PhoneService',
            'InternetService',
            'OnlineSecurity',
            'OnlineBackup',
            'DeviceProtection',
            'TechSupport',
            'StreamingTV',
            'StreamingMovies']

df['TotalServices'] = df[services].apply(
    lambda x: (x=="Yes").sum(),
    axis=1
)
```

```python
plt.figure(figsize=(9,5))

ax = sns.countplot(
    data=df,
    x='TotalServices',
    hue='Churn',
    palette='flare'
)

for c in ax.containers:
    ax.bar_label(c)

plt.title("Number of Services Used vs Churn", fontsize=16, fontweight='bold')
plt.show()
```

"Customers using more services generally demonstrate stronger loyalty. Offering bundled services could help improve customer retention."

# **Business Recommendations**
- Encourage Month-to-Month customers to switch to yearly contracts.
- Offer discounts for customers with high monthly charges.
- Promote Tech Support and Online Security to reduce churn.
- Focus retention efforts on customers with less than one year of tenure.
- Provide loyalty rewards for long-term customers.
